This is the script that imports text and cleans it, then converts it to utf-8

```python
import requests
import unicodedata
from pathlib import Path
import numpy as np
```

```python
from src.text_process.cleantxt import CleanText

from src.text_process.bpe import Tokenizer

SHERLOCK_URL = "https://www.gutenberg.org/cache/epub/1661/pg1661.txt"

TEXT = requests.get(SHERLOCK_URL).text
assert isinstance(TEXT, str)
```

```python
encoder_filepath = Path("/home/seba/Projects/NoTorchGrad/datasets/encoded_text.npy")
enc_txt_h_path = Path("/home/seba/Projects/NoTorchGrad/datasets/encoded_text_h.npy")
encoder_filepath.parent.mkdir(parents=True, exist_ok=True)

# clean the text
clean_text = CleanText.clean_text(TEXT)

# turn the cleaned text into bytes for training
encoded_text = clean_text.encode("utf-8")

tokenizer = Tokenizer()

# run the training on the text
tokenizer.train(data=encoded_text, vocab_size=1000) # 1000-256 merges
# encoded text
tokenizer.encoder(clean_text, save_to=encoder_filepath)
```
